```python
import sqlite3
from datetime import datetime

print("Cloud-Based Bus Pass System")
print("System initialized successfully!")
```

```python


connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE bus_passes (
    pass_id INTEGER PRIMARY KEY AUTOINCREMENT,
    passenger_name TEXT NOT NULL,
    bus_route TEXT NOT NULL,
    fare REAL NOT NULL,
    booking_date TEXT NOT NULL
)
""")

connection.commit()

print("Bus Pass Database created successfully!")
print("Table: bus_passes")
```

```python
# Bus routes and fixed fares

routes = {
    "Chennai-Madurai": 450,
    "Chennai-Coimbatore": 500,
    "Madurai-Tirunelveli": 250,
    "Tirunelveli-Tenkasi": 150
}

def book_bus_pass(passenger_name, bus_route):
    # Check whether the route exists
    if bus_route not in routes:
        return "Booking Failed: Invalid bus route."

    fare = routes[bus_route]

    # Add booking to database
    cursor.execute("""
    INSERT INTO bus_passes
    (passenger_name, bus_route, fare, booking_date)
    VALUES (?, ?, ?, ?)
    """, (
        passenger_name,
        bus_route,
        fare,
        datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    ))

    connection.commit()

    pass_id = cursor.lastrowid

    return (
        f"Booking Successful!\n"
        f"Pass ID: {pass_id}\n"
        f"Passenger: {passenger_name}\n"
        f"Route: {bus_route}\n"
        f"Fare: ₹{fare}"
    )

print(book_bus_pass("Ayisha", "Tirunelveli-Tenkasi"))
```

```python


def secure_booking(passenger_name, bus_route):

    cursor.execute("""
    SELECT * FROM bus_passes
    WHERE passenger_name = ? AND bus_route = ?
    """, (passenger_name, bus_route))

    existing_booking = cursor.fetchone()

    if existing_booking:
        return "Booking Failed: Duplicate booking detected!"

    return book_bus_pass(passenger_name, bus_route)


# Test duplicate booking
print(secure_booking("Rahul", "Chennai-Madurai"))
print()
print(secure_booking("Rahul", "Chennai-Madurai"))
```

```python
# Display all booked bus passes

def view_all_passes():
    cursor.execute("SELECT * FROM bus_passes")
    passes = cursor.fetchall()

    print("===== BUS PASS BOOKINGS =====")

    for bus_pass in passes:
        print(f"""
Pass ID       : {bus_pass[0]}
Passenger     : {bus_pass[1]}
Bus Route     : {bus_pass[2]}
Fare          : ₹{bus_pass[3]}
Booking Date  : {bus_pass[4]}
------------------------------
""")

view_all_passes()
```

```python
# Simulate dynamic server provisioning for high traffic

def server_provisioning(number_of_users):
    if number_of_users <= 100:
        servers = 1
    elif number_of_users <= 500:
        servers = 2
    elif number_of_users <= 1000:
        servers = 4
    else:
        servers = 6

    print("===== SERVER PROVISIONING =====")
    print("Number of users:", number_of_users)
    print("Servers required:", servers)
    print("System status: Scalable and Reliable")


# Test different traffic levels
server_provisioning(50)

print()

server_provisioning(750)

print()

server_provisioning(1500)
```

```python
# Final system testing

print("===== BUS PASS SYSTEM TESTING =====")

# Test 1: Valid booking
print("\nTest 1 - Valid Booking:")
print(secure_booking("Meena", "Chennai-Coimbatore"))

# Test 2: Duplicate booking
print("\nTest 2 - Duplicate Booking:")
print(secure_booking("Meena", "Chennai-Coimbatore"))

# Test 3: Invalid route
print("\nTest 3 - Invalid Route:")
print(secure_booking("Arun", "Chennai-Trichy"))

# Test 4: Display all bookings
print("\nTest 4 - Booking Records:")
view_all_passes()

print("\n===== ALL TESTS COMPLETED =====")
```
